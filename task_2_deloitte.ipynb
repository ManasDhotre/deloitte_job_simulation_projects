```python
# Importing Library
import pandas as pd
```

```python
# Load excel file 
input_file = r"C:\Users\Public\DELOITTE\data\Task 5 Equality Table.xlsx"
```

```python
df = pd.read_excel(input_file)
```

```python
print(df.columns.tolist())
```

```python
df.head()
```

```python
# Classify equality score
def equality_score(score):
    if score < -20 or score > 20:
        return "Highly Discriminative"
    elif -10 <= score <= 10:
        return "Fair"
    else:
        return "Unfair"
```

```python
df["Equality class"] = df["Equality Score"].apply(equality_score)
```
